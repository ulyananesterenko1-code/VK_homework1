# Задача 1. Развернуть часть массива

## Условие
Дан массив целых чисел. Необходимо повернуть (сдвинуть) справа налево часть массива, которая указана параметром `k`.
Решение должно работать за **линейное время** без дополнительных аллокаций.

**Пример:**
* Исходный массив: `[1, 2, 3, 4, 5, 6, 7]`
* `k = 3`
* Результат: `[5, 6, 7, 1, 2, 3, 4]`

**Что нужно сделать:**

Реализуйте функцию, которая выполняет циклический сдвиг части массива in-place.
Объясните, почему ваше решение работает за O(n) и не использует дополнительную
память.


```python
def reverse(sub_arr, start, end):
        while start < end:
            sub_arr[start], sub_arr[end] = sub_arr[end], sub_arr[start]
            start += 1
            end -= 1

arr = list(map(int, input("Введите элементы массива через пробел: ").split()))
k = int(input("Введите k (на сколько сдвинуть): "))

print("Исходный массив:", arr)

n = len(arr)
k = k % n

if k == 0:
  print("Масив после сдвега:", arr)

else :
  reverse(arr, 0, n - 1)
  reverse(arr, 0, k - 1)
  reverse(arr, k, n - 1)
  print("Масив после сдвега:", arr)
```

Мой алгоритм использует 3 разворота:
* Весь массив за O(n)
* Первые k элеменьов за O(k)
* Оставшиеся n-k элементов за O(n-k)

В итоге n+k+n-k=2n, константу можно отбросить => Сложность алгоритма = O(n)

# Задача 3. Слияние двух отсортированных массивов (без дополнительных аллокаций)

## Условие
* Дано два отсортированных массива.
* Первый массив имеет размер результирующего массива, а его конец заполнен нулями.
* Объединить массивы необходимо **без создания третьего массива**.

**Что нужно сделать:**

Напишите функцию, которая объединяет массивы, заполняя первый массив с конца. Поясните, почему заполнение с конца позволяет избежать перезаписи данных.


```python
arr1 = list(map(int, input("Введите элементы отсортированного результирующего массива через пробел: ").split()))
arr2 = list(map(int, input("Введите элементы отсортированного массива через пробел: ").split()))

m = int(input("Введите m (сколько реальных элементов в первом массиве): "))
n = len(arr2)

p1 = m-1
p2 = n-1
p = m+n-1

while p2 >= 0:
  if p1 >= 0 and arr1[p1] > arr2[p2]:
      arr1[p] = arr1[p1]
      p1 -= 1
  else:
      arr1[p] = arr2[p2]
      p2 -= 1
  p -= 1

print("Массив после слияния:", arr1)
```

Заполнение с конца позволяет избежать перезаписи, потому что мы используем нули в конце arr1 как свободное место. Мы сравниваем самые большие элементы и ставим наибольший в конец arr1.

# Задача 4. Сортировка массива из 0 и 1

## Условие
* Дан массив, содержащий только `0` и `1`.
* Напишите функцию, которая сортирует его так, чтобы все нули оказались в начале, а единицы — в конце.
* Решение должно быть **in-place**.

**Что нужно сделать:**

Реализуйте алгоритм, который проходит массив один раз и переставляет элементы без создания нового массива. Объясните, почему ваше решение имеет линейную сложность.

```python
arr = list(map(int, input("Введите элементы массива через пробел (элементами являются только 0 и 1): ").split()))

left = 0
right = len(arr) - 1

while left < right:
  if arr[left] == 0:
    left += 1
  elif arr[right] == 1:
    right -= 1
  else:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print("Отсортированный массив:", arr)
```

Сложность O(n), потому что указатели left и right проходят массив навстречу друг другу, и каждый элемент проверяется ровно один раз. Количество итераций цикла пропорционально длине массива n.

# Задача 5. Задача флага Нидерландов

## Условие
* Дан массив, состоящий из `0`, `1` и `2`.
* Необходимо отсортировать его за **линейное время**.

**Что нужно сделать:**

Реализуйте алгоритм трёх указателей (Dutch National Flag). Кратко опишите, как поддерживаются границы для 0, 1 и 2 во время прохода по массиву.

```python
arr = list(map(int, input("Введите элементы массива через пробел (элементами являются только 0, 1 и 2): ").split()))

p1 = 0
p2 = 0
p3 = len(arr) - 1

while p2 <= p3:
  if arr[p2] == 0:
     arr[p1], arr[p2] = arr[p2], arr[p1]
     p1 += 1
     p2 += 1
  elif arr[p2] == 1:
     p2 += 1
  else:
     arr[p2], arr[p3] = arr[p3], arr[p2]
     p3 -= 1

print("Отсортированный массив:", arr)
```

Как поддерживаются границы:
* [0 ... p1-1] — зона нулей.
* [p1 ... p2-1] — зона единиц.
* [p2 ... p3] — непроверенная часть.
* [p3+1 ... конец] — зона двоек.

Указатель p2 идет по непроверенной части. Если видим 0 — меняем с p1 и двигаем оба. Если 1 — просто двигаем p2. Если 2 — меняем с p3, двигаем только p3.

# Задача 6. Передвинуть чётные числа вперёд

## Условие
* Дан неотсортированный массив целых чисел.
* Необходимо перенести в начало массива все чётные числа, **сохраняя их относительный порядок**.

**Пример:**
Ввод: `[3, 2, 4, 1, 11, 8, 9]`
Вывод: `[2, 4, 8, 1, 11, 3, 9]`

**Что нужно сделать:**
Реализуйте стабильное перераспределение элементов, сохраняя порядок чётных чисел. Объясните, за счёт чего сохраняется относительный порядок элементов.

```python
arr = list(map(int, input("Введите элементы массива через пробел: ").split()))

j = 0
for i in range(len(arr)):
  if arr[i] % 2 == 0:
    arr[j], arr[i] = arr[i], arr[j]
    j += 1

print("Отсортированный массив:", arr)
```

Стабильность достигается благодаря тому, что мы не меняем местами два чётных числа между собой.

# Задача 7. Нули в конец

## Условие
В школе прошёл экзамен по математике. Несколько человек списали решения и получили 0 баллов.

* Дан массив оценок, среди которых есть `0`.
* Необходимо перенести все нули в конец массива, чтобы такие ученики оказались в конце списка.
* Порядок остальных элементов должен сохраняться.

**Что нужно сделать:**
Реализуйте алгоритм, который за один проход переносит все ненулевые элементы вперёд. Поясните, почему ваше решение является стабильным и работает за \(O(n)\).


```python
arr = list(map(int, input("Введите элементы массива через пробел: ").split()))

j = 0
for i in range(len(arr)):
  if arr[i] != 0:
    arr[j], arr[i] = arr[i], arr[j]
    j += 1

print("Отсортированный массив:", arr)
```

* Стабильность достигается благодаря тому, что мы не меняем местами два ненулевых числа между собой.
* Решение работает за O(n) поскольку мы проходимся по массиву ровно 1 раз
